```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests

load_dotenv()
```

```python
@tool
def get_conversion_factor(base_currency: str, target_currency: str)-> float:
    """This functon fetches the currency convertion factor btw base currency and taget currency"""

    url = f'https://v6.exchangerate-api.com/v6/744b1274ea4ec48edbea15f6/pair/{base_currency}/{target_currency}'

    response = requests.get(url)

    return response.json()
```

```python
get_conversion_factor.invoke({'base_currency':'USD','target_currency': 'INR'})
```

```python
@tool
def converter(base_currency: int,conversion_rate: float) -> float:
    """Given a currency converter rate this function calculate the target currency value from a given base currency value"""
    return base_currency*conversion_rate
```

```python
converter.invoke({'base_currency': 10, 'conversion_rate':95.6063 })
```
